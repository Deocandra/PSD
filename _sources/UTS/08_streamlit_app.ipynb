# 08. Deployment Aplikasi Web Dashboard: Streamlit
## Otomatisasi Pembuatan `app.py` dan `requirements.txt`

Notebook ini menghasilkan kode program aplikasi web dashboard interaktif **Streamlit (`app.py`)** dan daftar dependensi **`requirements.txt`**.
Aplikasi ini menyatukan seluruh capaian proyek dalam 6 tab:
1. **Business Understanding** (menampilkan teks `index.md` & ringkasan tujuan)
2. **Data Understanding & Preparation** (tabel piksel, spectral signatures, agregasi poligon)
3. **Modeling & Evaluasi (SVM)** (metrik akurasi, Macro F1, classification report, confusion matrix)
4. **Peta Aktual (Ground Truth)** (peta interaktif poligon data aktual)
5. **Prediksi Model (SVM)** (peta evaluasi prediksi, filter benar/salah, tombol unduh CSV)
6. **Prediksi Raster Jawa Timur** (peta klasifikasi skala penuh seluruh Jawa Timur & tabel luas per kelas)

```python
from pathlib import Path

BASE_DIR = Path.cwd()
APP_FILE = BASE_DIR / "app.py"
REQ_FILE = BASE_DIR / "requirements.txt"

print("Direktori target:", BASE_DIR.resolve())
```

### 1. Menulis Berkas Dependensi `requirements.txt`

```python
requirements_content = '''streamlit>=1.50.0
streamlit-folium>=0.20.0
folium>=0.16.0
pandas>=2.2.0
numpy>=1.26.0
scikit-learn>=1.4.0
geopandas>=0.14.0
rasterio>=1.3.0
joblib>=1.3.0
matplotlib>=3.8.0
seaborn>=0.13.0
Pillow>=10.0.0
'''

with open(REQ_FILE, "w", encoding="utf-8") as f:
    f.write(requirements_content.strip())

print(f"File requirements.txt tersimpan di: {REQ_FILE.resolve()}")
```

### 2. Memeriksa Ketersediaan File `app.py`

```python
if APP_FILE.exists():
    print(f"Berkas app.py ditemukan: {APP_FILE.resolve()} ({APP_FILE.stat().st_size:,} bytes)")
else:
    print(f"Peringatan: Berkas app.py belum ada di {APP_FILE.resolve()}")
```

### 3. Pengujian Aplikasi Menggunakan `streamlit.testing.v1.AppTest`
Menjalankan pengujian sintaksis dan eksekusi headless menggunakan modul pengujian resmi Streamlit.

```python
from streamlit.testing.v1 import AppTest

print("Menjalankan AppTest untuk memvalidasi kestabilan app.py...")
at = AppTest.from_file(str(APP_FILE), default_timeout=30)
at.run()

if at.exception:
    print("Ditemukan exception pada pengujian:")
    for ex in at.exception:
        print(ex)
    raise RuntimeError("AppTest gagal!")
else:
    print("UJI APLIKASI STREAMLIT BERHASIL TANPA EXCEPTION!")
    print(f"Jumlah tab terdeteksi: {len(at.tabs)}")
```
